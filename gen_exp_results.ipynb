```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import re

folder = "./experiment_results/generalization_exp"
files = os.listdir(folder)

dfs = []

for f in files:
    print(f)
    if f.endswith(".csv"):
        # extract the number at the end of the filename
        # e.g. "generalization_exp7.csv" → "7"
        # run_num = re.findall(r'\d+', f)[-1]

        df = pd.read_csv(os.path.join(folder, f))
        
        # replace the run column with the extracted run number
        # df["run"] = int(run_num)

        dfs.append(df)

# merge everything
merged = pd.concat(dfs, ignore_index=True)

```

```python
df_mean = (
    merged.groupby(['dataset', 'class_name', 'removal', 'pqr'], as_index=False)
      .agg(
          all_data_recall_mean=('all_data_recall', 'mean'),
          all_data_recall_std=('all_data_recall', 'std'),
          all_data_f1_mean=('all_data_f1', 'mean'),
          all_data_f1_std=('all_data_f1', 'std'),
          all_data_precision_mean=('all_data_precision', 'mean'),
          all_data_precision_std=('all_data_precision', 'std'),
          all_data_accuracy_mean=('all_data_accuracy', 'mean'),
          all_data_accuracy_std=('all_data_accuracy', 'std'),
          all_data_jaccard_mean=('all_data_jaccard', 'mean'),
          all_data_jaccard_std=('all_data_jaccard', 'std'),
          removed_data_recall_mean=('removed_data_recall', 'mean'),
          removed_data_recall_std=('removed_data_recall', 'std'),
          removed_data_f1_mean=('removed_data_f1', 'mean'),
          removed_data_f1_std=('removed_data_f1', 'std'),
          removed_data_precision_mean=('removed_data_precision', 'mean'),
          removed_data_precision_std=('removed_data_precision', 'std'),
          removed_data_accuracy_mean=('removed_data_accuracy', 'mean'),
          removed_data_accuracy_std=('removed_data_accuracy', 'std'),
          removed_data_jaccard_mean=('removed_data_jaccard', 'mean'),
          removed_data_jaccard_std=('removed_data_jaccard', 'std'),
          
      )
      .reset_index()
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np


def plot_each_metric_separately(df_mean, data_type="all_data", output_prefix="metric_plot"):
    """
    Creates and saves one PDF per metric (recall, precision, accuracy, jaccard)
    showing mean ± std vs removal rate.
    """
    metrics = ["recall", "precision", "accuracy", "jaccard"]

    for metric in metrics:

        mean_col = f"{data_type}_{metric}_mean"
        std_col = f"{data_type}_{metric}_std"

        # Prepare data
        df_plot = (
            df_mean
            .groupby(['dataset', 'removal', 'pqr'], as_index=False)[[mean_col, std_col]]
            .mean()
        )

        # Create SINGLE figure per metric
        plt.figure(figsize=(7, 5))

        for pqr_method, group in df_plot.groupby("pqr"):
            plt.plot(
                group["removal"],
                group[mean_col],
                marker="o",
                label=pqr_method,
                linewidth=3
            )

            plt.fill_between(
                group["removal"],
                group[mean_col] - group[std_col],
                group[mean_col] + group[std_col],
                alpha=0.2
            )

        #plt.title(metric.replace("_", " ").title(), fontsize=22, weight="bold")
        plt.xlabel("Removal Rate", fontsize=28)
        plt.yticks(fontsize=25)
        plt.xticks(fontsize=25)
        plt.ylabel(metric.replace("_", " ").title(), fontsize=28,weight="bold")
        plt.grid(True, alpha=0.3)
        # plt.legend(title="PQR Method", fontsize=11)

        # Save as PDF (one file per metric)
        filename = f"{output_prefix}_{metric}.pdf"
        plt.savefig(filename, bbox_inches='tight')
        plt.close()

        print(f"Saved: {filename}")
```

```python
# For all data
plot_each_metric_separately(df_mean, data_type="all_data")
```
