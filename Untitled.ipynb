```python
import pandas as pd
import numpy as np
import umap
import matplotlib.pyplot as plt


import json
import numpy as np
import pandas as pd
import pickle

def load_csv(filepath):
    return pd.read_csv(filepath)

def load_mapping(filepath):
    with open(filepath, 'rb') as f:
        return pickle.load(f)

def entity_vectors(learning_problem, pqr, label_na=-1, entity_filter=False):
    # File paths
    base_path = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}"
    entities_file = f"{base_path}/DeCaL_entity_embeddings.csv"
    entity_map_file = f"{base_path}/entity_to_idx.p"
    lp_json_file = "./LPs/Family/lps.json"

    # Load entity embeddings and map
    entities_df = load_csv(entities_file)  # First column is URI
    entity_map = load_mapping(entity_map_file)

    # Load learning problem examples
    with open(lp_json_file, "r") as f:
        problems = json.load(f)["problems"]

    if learning_problem not in problems:
        raise ValueError(f"Learning problem '{learning_problem}' not found in LP.json.")

    pos_uris = set(problems[learning_problem]["positive_examples"])
    neg_uris = set(problems[learning_problem]["negative_examples"])

    # Build entity vectors and labels
    vectors = []
    labels = []
    uris = []

    for _, row in entities_df.iterrows():
        uri = row[0]
        vec = row[1:].to_numpy()

        if uri in pos_uris:
            label = 1
        elif uri in neg_uris:
            label = 0
        else:
            label = label_na

        if not entity_filter or label != label_na:
            uris.append(uri)
            vectors.append(vec)
            labels.append(label)

    vectors = np.array(vectors)
    labels = np.array(labels)

    return vectors, labels, uris



# Step 1: Load CSV files
def load_csv(file_path):
    """Load a CSV file into a DataFrame."""
    return pd.read_csv(file_path)


# Step 1: Load CSV files
def load_mapping(file_path):
    """Load a .p file into a DataFrame."""
    return pd.read_pickle(file_path)


def clifford_product_matrix(X):
    """
    Computes the pairwise Clifford product <u,v>_p,q,r for all vectors in X.

    Parameters:
    - X: numpy array of shape (N, 48), each row is a concatenated triple (u_p, u_q, u_r)

    Returns:
    - result: numpy array of shape (N, N) with Clifford products for each (i,j) pair
    """
    # assert X.shape[1] == 48, "Expected 48-dimensional vectors"
    assert X.shape[1] == 32, "Expected 16-dimensional vectors"
    # Split into three 16-dimensional components
    X_p = X[:, :32]  # Shape: (N, 16)
    # X_q = X[:, 16:32]
    # X_r = X[:, 32:]

    # Compute pairwise dot products for each part
    dot_p = X_p @ X_p.T  # Shape: (N, N)
    # dot_q = X_q @ X_q.T
    # dot_r = X_r @ X_r.T

    # Sum the components to get the Clifford product
    return dot_p  # + dot_q + dot_r


def clifford_to_distance(S):
    """
    Converts a similarity matrix (e.g., Clifford product) to a distance matrix.

    Parameters:
    - S: similarity matrix (N x N)

    Returns:
    - D: distance matrix (N x N)
    """
    max_sim = np.max(S)
    return max_sim - S

def get_concat_vectors(pqr):
    entities_file = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}/DeCaL_entity_embeddings.csv"
    relations_file = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}/DeCaL_relation_embeddings.csv"
    entity_map_file = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}/entity_to_idx.p"
    relation_map_file = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}/relation_to_idx.p"

    # Load data
    entities_df = load_csv(entities_file)
    relations_df = load_csv(relations_file)
    entity_map = load_mapping(entity_map_file)
    relation_map = load_mapping(relation_map_file)

    # Convert entity and relation vectors to numpy arrays
    entity_vectors = entities_df.iloc[:, 1:].to_numpy()  # Assuming vectors start from column 1
    relation_vectors = relations_df.iloc[:, 1:].to_numpy()  # Assuming vectors start from column 1

    # Example triples
    # files = 'test'
    # triples = np.load(f'./KGs_Family_family-benchmark_rich_background_owl_{pqr}/train_set.npy')

    # Process triples
    # vectors, labsq = process_triples(triples, entity_vectors, relation_vectors)
    return np.concatenate([entity_vectors, relation_vectors])
```

```python
import matplotlib.pyplot as plt
import umap
import numpy as np

def plot_lp_embedding_grid(lp, pqrs, label_na=-1, save_path=None, show=True):
    fig, axes = plt.subplots(2, 4, figsize=(20, 10))
    axes = axes.flatten()

    # Define consistent colormap and label names
    label_colors = {
        1: 'green',
        0: 'red',
        -1: 'gray'
    }

    label_names = {
        1: 'Positive',
        0: 'Negative',
        -1: 'Unlabeled'
    }

    for i, pqr in enumerate(pqrs):
        vectors, labels, _ = entity_vectors(lp, pqr, label_na=label_na, entity_filter=True)

        cliff_dist = clifford_to_distance(clifford_product_matrix(vectors))
        f = umap.UMAP(metric='precomputed', random_state=42).fit(cliff_dist)
        f_mapped = f.transform(cliff_dist)

        ax = axes[i]
        for label_val in np.unique(labels):
            mask = labels == label_val
            ax.scatter(f_mapped[mask, 0], f_mapped[mask, 1],
                       c=label_colors[label_val],
                       label=label_names[label_val],
                       s=25, alpha=0.7, edgecolors='k')

        ax.set_title(f"{pqr}", fontsize=14)
        ax.set_xticks([])
        ax.set_yticks([])

    # Unified legend outside the subplots
    handles = [plt.Line2D([0], [0], marker='o', color='w',
               markerfacecolor=label_colors[l], label=label_names[l],
               markersize=10, markeredgecolor='k') for l in label_names]
    fig.legend(handles=handles, title="Label", loc='lower center', ncol=3, fontsize=12, title_fontsize=13)

    fig.suptitle(f"UMAP Embeddings for Learning Problem: {lp}", fontsize=20)
    plt.tight_layout(rect=[0, 0.05, 1, 0.95])

    if save_path:
        plt.savefig(save_path, dpi=300)
    if show:
        plt.show()
    plt.close()




```

```python
from sklearn.neighbors import KNeighborsClassifier
def leave_one_out_knn_dist_err(dists, labs, n_neighbors=5):
    nn = KNeighborsClassifier(n_neighbors=n_neighbors, metric = 'precomputed')
    nn.fit(dists, labs) 
    
    unique_l = np.unique(labs)
    errs = 0
    # calculate the leave one out nearest neighbour error for each point
    neighs = nn.kneighbors(return_distance=False)
    neigh_labs = labs[neighs]
    counts_cl = np.zeros([labs.shape[0], unique_l.shape[0]])
    for i in range(unique_l.shape[0]):
        counts_cl[:,i] = np.sum(neigh_labs  == unique_l[i], 1)
    
    pred_labs = unique_l[counts_cl.argmax(1)]
    
    # calculate the prediction error
    return sum(pred_labs != labs)/labs.shape[0]
```

```python
from collections import defaultdict
import numpy as np

# Initialize dictionary to store errors per pqr
pqr_errors = defaultdict(list)

lps = list(problems.keys())
pqrs = ["0_0_0", "0_0_1", "0_1_0", "0_1_1", "1_0_0", "1_0_1", "1_1_0", "1_1_1"]

# Loop through each learning problem and pqr
for lp in lps:
    print(f"Learning problem: {lp}")
    for pqr in pqrs:
        vectors, labels, _ = entity_vectors(lp, pqr, label_na=-1, entity_filter=True)

        # Skip if not enough labeled examples
        if len(np.unique(labels)) < 2:
            print(f"  Skipping {pqr}: not enough label diversity.")
            continue

        cliff_dist = clifford_to_distance(clifford_product_matrix(vectors))
        error = leave_one_out_knn_dist_err(cliff_dist, labels)
        print(f"  Error for {pqr}: {error:.4f}")

        pqr_errors[pqr].append(error)

# Compute average error for each pqr
print("\nAverage Errors per PQR Setting:")
avg_errors = {}
for pqr in pqrs:
    errors = pqr_errors[pqr]
    if errors:
        avg_error = np.mean(errors)
        avg_errors[pqr] = avg_error
        print(f"{pqr}: {avg_error:.4f}")
    else:
        print(f"{pqr}: No valid data")

# Optional: print best setting
best_pqr = min(avg_errors, key=avg_errors.get)
print(f"\nBest PQR setting based on average error: {best_pqr} ({avg_errors[best_pqr]:.4f})")
```

```python
def entity_vectors_disjoint_lps(lp_list, pqr):
    """
    For a given set of LPs and a PQR setting, return entity vectors of disjoint positive examples.
    Each learning problem is treated as a class. Only positive examples are considered.
    
    Returns:
        vectors: np.ndarray of shape (n, d)
        labels: list of LP names
        uris: list of entity URIs
    """
    import json

    # File paths
    base_path = f"./KGs_Family_family-benchmark_rich_background_owl_{pqr}"
    entities_file = f"{base_path}/DeCaL_entity_embeddings.csv"
    entity_map_file = f"{base_path}/entity_to_idx.p"
    lp_json_file = "./LPs/Family/lps.json"

    # Load entity embeddings
    entities_df = load_csv(entities_file)  # Assume first col is URI
    entity_vecs = {row[0]: row[1:].to_numpy() for _, row in entities_df.iterrows()}

    # Load learning problems
    with open(lp_json_file, "r") as f:
        problems = json.load(f)["problems"]

    assigned_uris = set()
    vectors = []
    labels = []
    uris = []

    for lp in lp_list:
        if lp not in problems:
            print(f"Skipping unknown LP: {lp}")
            continue

        pos_uris = problems[lp]["positive_examples"]
        for uri in pos_uris:
            if uri not in assigned_uris and uri in entity_vecs:
                vectors.append(entity_vecs[uri])
                labels.append(lp)
                uris.append(uri)
                assigned_uris.add(uri)

    vectors = np.array(vectors)
    unique_labels = list(set(labels))
    label_to_num = {label: idx for idx, label in enumerate(unique_labels)}
    numeric_labels = [label_to_num[label] for label in labels]
    
    return vectors, numeric_labels,labels, uris
```
